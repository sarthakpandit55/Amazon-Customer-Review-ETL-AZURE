### Reading Data

```python
df = spark.read \
    .format("csv") \
    .option("header", "true") \
    .option("multiLine", "true") \
    .option("quote", '"') \
    .option("escape", '"') \
    .option("mode", "PERMISSIVE") \
    .load("abfss://bronze@customerreviewdatalake.dfs.core.windows.net/raw_data")
```
